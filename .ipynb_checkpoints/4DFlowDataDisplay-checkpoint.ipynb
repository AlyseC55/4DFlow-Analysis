```python
import pydicom
from pathlib import Path
import pandas as pd

main_folder = Path(r"\\pom\pmcommon\_PM.Private\CSIC\Moulin\4D Flow Data")

records = []

for patient_dir in sorted(main_folder.iterdir()):
    if not patient_dir.is_dir():
        continue
    patient_id = patient_dir.name

    for sequence_dir in sorted(patient_dir.iterdir()):
        if not sequence_dir.is_dir():
            continue
        sequence_id = sequence_dir.name

        dcm_files = list(sequence_dir.glob("*.dcm"))
        for dcm_file in dcm_files:
            try:
                ds = pydicom.dcmread(dcm_file)
                records.append({
                    "patient_folder": patient_id,
                    "sequence_folder": sequence_id,
                    "filename": dcm_file.name,
                    "path": str(dcm_file),
                    "PatientID_tag": getattr(ds, "PatientID", None),
                    "SeriesDescription": getattr(ds, "SeriesDescription", None),
                    "Modality": getattr(ds, "Modality", None),
                    "InstanceNumber": getattr(ds, "InstanceNumber", None),
                    "rows": getattr(ds, "Rows", None),
                    "cols": getattr(ds, "Columns", None),
                })
            except Exception as e:
                print(f"Skipped {dcm_file} due to error: {e}")
df = pd.DataFrame(records)
df.head()
```
